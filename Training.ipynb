```python
import pandas as pd
data=pd.read_csv("Mall_Customers.csv")
```

```python
X=data[["Age","Annual Income (k$)",	"Spending Score (1-100)"]]
```

```python
import os
os.environ["OMP_NUM_THREARDS"]="1"
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
X_transformes=scaler.fit_transform(X)
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score, make_scorer
```

```python
best_score=-1
best_k=None
best_model=None
for k in range(2,9):
    model=KMeans(n_clusters=k, init='k-means++', random_state=42)
    labels=model.fit_predict(X_transformes)
    if len(set(labels)) > 1:
        score=silhouette_score(X_transformes, labels)
        print(f"k={k} -> Silhouette Score = {round(score,3)}")
        if score > best_score:
            best_score=score
            best_k = k
            best_model=model
```

```python
print("\nMeilleur modele selectionne")
print(f"Nombre optimal de clusters (k): {best_k}")
print("Score de Silhouette optimal: ",round(best_score,3))
```

```python
sil=silhouette_score(X_transformes, labels)
db=davies_bouldin_score(X_transformes, labels)
ch=calinski_harabasz_score(X_transformes, labels)
print("\nEvaluation Finale du Clustering")
print("Silhouette Score: ", round(sil,3))
print("Davies Bouldin Score: ", round(db,3))
print("Calinski-Harabasz Score: ", round(ch,3))
```

```python
from sklearn.cluster import AgglomerativeClustering, DBSCAN
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score
import pandas as pd

# Dictionnaire pour stocker les résultats et remplir facilement votre tableau
comparaison_resultats = []

# --- 1. Agglomerative Clustering (on fixe n_clusters=6 pour comparer à la baseline) ---
agg_model = AgglomerativeClustering(n_clusters=6)
agg_labels = agg_model.fit_predict(X_transformes)

comparaison_resultats.append({
    "Algorithme": "Agglomerative",
    "Nb. clusters": len(set(agg_labels)),
    "Silhouette ↑": silhouette_score(X_transformes, agg_labels),
    "Davies-Bouldin ↓": davies_bouldin_score(X_transformes, agg_labels),
    "Calinski-Harabasz ↑": calinski_harabasz_score(X_transformes, agg_labels)
})

# --- 2. DBSCAN (ajustez eps et min_samples si nécessaire selon la densité) ---
dbscan_model = DBSCAN(eps=0.5, min_samples=5)
dbscan_labels = dbscan_model.fit_predict(X_transformes)
n_clusters_db = len(set(dbscan_labels)) - (1 if -1 in dbscan_labels else 0)

if n_clusters_db > 1:
    comparaison_resultats.append({
        "Algorithme": "DBSCAN",
        "Nb. clusters": n_clusters_db,
        "Silhouette ↑": silhouette_score(X_transformes, dbscan_labels),
        "Davies-Bouldin ↓": davies_bouldin_score(X_transformes, dbscan_labels),
        "Calinski-Harabasz ↑": calinski_harabasz_score(X_transformes, dbscan_labels)
    })
else:
    print("DBSCAN a généré trop de bruit ou un seul cluster. Ajustez 'eps'.")

# --- 3. Gaussian Mixture Model (GMM) ---
gmm_model = GaussianMixture(n_components=6, random_state=42)
gmm_labels = gmm_model.fit_predict(X_transformes)

comparaison_resultats.append({
    "Algorithme": "Gaussian Mixture",
    "Nb. clusters": 6,
    "Silhouette ↑": silhouette_score(X_transformes, gmm_labels),
    "Davies-Bouldin ↓": davies_bouldin_score(X_transformes, gmm_labels),
    "Calinski-Harabasz ↑": calinski_harabasz_score(X_transformes, gmm_labels)
})

# --- Affichage sous forme de tableau propre ---
df_comparaison = pd.DataFrame(comparaison_resultats)
print("\n--- TABLEAU COMPARATIF DES MODELES ---")
display(df_comparaison)
```

```python
# Comparaison de plusieurs valeurs de eps pour améliorer la segmentation DBSCAN

from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score
import numpy as np

for eps in [0.5, 0.7, 0.9, 1.1, 1.3, 1.5]:
    model = DBSCAN(eps=eps, min_samples=5)
    labels = model.fit_predict(X_transformes)

    n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
    n_bruit = np.sum(labels == -1)

    if n_clusters >= 2:
        masque = labels != -1
        score = silhouette_score(
            X_transformes[masque],
            labels[masque]
        )
        print(
            f"eps={eps} | Clusters={n_clusters} | "
            f"Bruit={n_bruit} | Silhouette sans bruit={score:.3f}"
        )
    else:
        print(
            f"eps={eps} | Clusters={n_clusters} | "
            f"Bruit={n_bruit} | Silhouette non calculable"
        )

```

```python
# Recherche des paramètres DBSCAN offrant une segmentation exploitable

from sklearn.cluster import DBSCAN
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)
import numpy as np
import pandas as pd

resultats_dbscan = []

for min_samples in [3, 4, 5, 6, 8, 10]:
    for eps in [0.3, 0.4, 0.5, 0.6, 0.7, 0.8]:

        modele = DBSCAN(
            eps=eps,
            min_samples=min_samples
        )

        labels = modele.fit_predict(X_transformes)

        # Exclusion des points de bruit identifiés par -1
        masque = labels != -1
        labels_clusters = labels[masque]
        donnees_clusters = X_transformes[masque]

        nb_clusters = len(set(labels_clusters))
        nb_bruit = np.sum(~masque)
        proportion_bruit = nb_bruit / len(labels)

        # Calcul des métriques si au moins deux clusters
        # et au moins deux observations par cluster au total
        if nb_clusters >= 2 and len(set(labels_clusters)) < len(labels_clusters):

            silhouette = silhouette_score(
                donnees_clusters,
                labels_clusters
            )

            davies = davies_bouldin_score(
                donnees_clusters,
                labels_clusters
            )

            calinski = calinski_harabasz_score(
                donnees_clusters,
                labels_clusters
            )

            resultats_dbscan.append({
                "eps": eps,
                "min_samples": min_samples,
                "Nb. clusters": nb_clusters,
                "Nb. bruit": nb_bruit,
                "Bruit (%)": round(proportion_bruit * 100, 1),
                "Silhouette": round(silhouette, 3),
                "Davies-Bouldin": round(davies, 3),
                "Calinski-Harabasz": round(calinski, 3)
            })

# Affichage des configurations exploitables
df_dbscan = pd.DataFrame(resultats_dbscan)

if not df_dbscan.empty:
    df_dbscan = df_dbscan.sort_values(
        by="Silhouette",
        ascending=False
    )

    display(df_dbscan)
else:
    print(
        "Aucune configuration n'a produit au moins deux "
        "clusters exploitables. Il faudra élargir la recherche."
    )

```

```python
# Évaluation correcte de DBSCAN en excluant les observations considérées comme du bruit

import numpy as np
from sklearn.cluster import DBSCAN
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)

# Entraîner DBSCAN avec les paramètres retenus provisoirement
dbscan_final = DBSCAN(eps=0.5, min_samples=5)
labels_dbscan = dbscan_final.fit_predict(X_transformes)

# Identifier les clients qui ne sont pas considérés comme du bruit
masque = labels_dbscan != -1
donnees_sans_bruit = X_transformes[masque]
labels_sans_bruit = labels_dbscan[masque]

# Calculer le nombre de clusters et de clients considérés comme du bruit
nb_clusters = len(set(labels_sans_bruit))
nb_bruit = int(np.sum(labels_dbscan == -1))
pourcentage_bruit = nb_bruit / len(labels_dbscan) * 100

# Afficher les informations de la segmentation
print("Nombre total de clients :", len(labels_dbscan))
print("Nombre de clusters :", nb_clusters)
print("Nombre de clients considérés comme du bruit :", nb_bruit)
print("Pourcentage de bruit :", round(pourcentage_bruit, 2), "%")

# Calculer les métriques uniquement si au moins deux clusters sont exploitables
if nb_clusters >= 2:
    print("Silhouette :", round(
        silhouette_score(donnees_sans_bruit, labels_sans_bruit), 3
    ))
    print("Davies-Bouldin :", round(
        davies_bouldin_score(donnees_sans_bruit, labels_sans_bruit), 3
    ))
    print("Calinski-Harabasz :", round(
        calinski_harabasz_score(donnees_sans_bruit, labels_sans_bruit), 3
    ))
else:
    print("Les métriques de comparaison ne peuvent pas être calculées.")

```
